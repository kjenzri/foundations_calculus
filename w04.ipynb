```python
def f(x):
    return x**2
```

```python
def f_derivative(x, h):
    return (f(x+h)-f(x))/h
```

```python
input = [-1e-1, -1e-2, -1e-3, -1e-4, 1e-4, 1e-3, 1e-2, 1e-1]
ouput = [f_derivative(3, h) for h in input]

print(ouput)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_near = np.linspace(-5, 10, 200)
plt.plot(x_near, x_near**2, label="f(x) = x²")

for h in [0.5, -0.5]:
    slope = f_derivative(3, h)
    secant = f(3) + slope * (x_near - 3)
    plt.plot(x_near, secant, label=f"Secant, h={h}")

tangent = f(3) + 6 * (x_near - 3)
plt.plot(x_near, tangent, label="Tangent")

plt.legend()
plt.show()

```

```python
import math
def function(x):
    return math.exp(x)

def analytic_derivative(x):
    return math.exp(x)

def finite_difference(x, h):
    return (function(x+h)-function(x))/h

h_values = [-1e-1, 1e-1, -1e-2, 1e-2, -1e-3, 1e-3, -1e-4, 1e-4]

near_0 = [(analytic_derivative(0), finite_difference(0, h)) for h in h_values]
near_0 = [(a, f, abs(a-f)) for a, f in near_0]
near_1 = [(analytic_derivative(1), finite_difference(1, h)) for h in h_values]
near_1 = [(a, f, abs(a-f)) for a, f in near_1]

print(near_0)
print(near_1)
```

```python
import math

def FV(r):
    return 1000 * math.exp(2 * r)

r = 0.05
h = 0.0001  # 1 bp

analytic = 2000 * math.exp(2 * r)
finite_difference = (FV(r + h) - FV(r)) / h

print("Analytic sensitivity:", analytic)
print("1 bp finite-difference sensitivity:", finite_difference)
print("Analytic 1 bp change:", analytic * h)
print("Revalued 1 bp change:", FV(r + h) - FV(r))
```
