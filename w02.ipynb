```python
def f(x):
    return 2*x+1

inputs = list(range(-10, 10, 1))
for x in inputs:
    result = f(x)
    #print(result)
```

```python
def g(x):
    return x**2

inputs = list(range(-10, 10, 1))
for x in inputs:
    result = g(x)
    #print(result)
```

```python
def f_g(x):
    y = g(x)
    return f(y)

def g_f(x):
    y = f(x)
    return g(y)

print(f_g(2))
print(g_f(2))
```

```python
from math import exp
import math
def continuous_fv(pv, r, t):
    return pv * exp(r * t)
actual = continuous_fv(100, 0.05, 1)
assert math.isclose(actual, 105.127, abs_tol=0.001)
```

```python
def discret_fv(pv, r, t):
    return pv * (1 + r) ** t

assert discret_fv(100, 0.05, 1) == 105
```

```python
def effective_to_continuous(r):
    return math.log(1+r)
```

```python
def continuous_to_effective(r_c):
    return math.exp(r_c) - 1
```

```python
effective = 0.05
assert math.isclose(effective, continuous_to_effective(effective_to_continuous(effective)))
```
