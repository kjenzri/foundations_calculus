```python
from math import exp
import math
def continuous_fv(pv, r, t):
    return pv * exp(r * t)
actual = continuous_fv(100, 0.05, 1)
assert math.isclose(actual, 105.127, abs_tol=0.001)
```

```python
def discrete_fv(pv, r, t):
    return pv * (1 + r) ** t

assert discrete_fv(100, 0.05, 1) == 105
```

```python
def effective_to_continuous(r):
    return math.log(1+r)
```

```python
def continuous_to_effective(r_c):
    return math.exp(r_c) - 1
```

```python
effective = 0.05
assert math.isclose(effective, continuous_to_effective(effective_to_continuous(effective)))
```

```python
def time_to_target_discrete(fv, pv, r):
    return math.log(fv/pv)/math.log(1+r)

expected = 9.0065
actual = time_to_target_discrete(200, 100, 0.08)

assert math.isclose(actual, expected, abs_tol=0.0001)
```

```python
def time_to_target_continuous(fv, pv, r_c):
    return math.log(fv/pv)/r_c

expected = 8.6643
actual = time_to_target_continuous(200, 100, 0.08)

assert math.isclose(actual, expected, abs_tol=0.0001)
```
