```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 2, 300)
y = (x**2 - 1) / (x - 1)

# Create a gap at x = 1
y[np.isclose(x, 1)] = np.nan

plt.plot(x, y, label="f(x)")

# Open circle at the missing point: (1, 2)
plt.plot(1, 2, marker="o", markersize=8,
         markerfacecolor="white", markeredgecolor="red",
         label="hole at x = 1")

plt.axvline(1, color="red", linestyle="--", alpha=0.5)
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.legend()
plt.show()
```

```python
pv = 1000
r = 0.05
t = 2

n = np.array([1,2,4,12,52,365,10000, 100000, 10000000000])
fv = pv*(1+r/n)**(n*t)

np.set_printoptions(precision=10, suppress=True)

display(fv)

import math
fv_continuous = pv * math.exp(r*t)
print(fv_continuous)

diff = fv - fv_continuous

display(diff)
```
