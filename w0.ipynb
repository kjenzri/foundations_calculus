```python
def future_value_discrete(pv,r,T):
    return pv * (1+r) ** T
fv = future_value_discrete(100, 0.05, 1)
assert fv == 105.0
```

```python
def present_value_discrete(fv,r,T):
    return fv / (1+r) ** T
pv = present_value_discrete(105, 0.05, 1)
assert pv == 100.0
```

```python
import math
def rate_discrete(fv,pv,T):
    return (fv / pv) ** (1/T) - 1
r = rate_discrete(121, 100, 2)
assert math.isclose(r, 0.10)
```

```python
def future_value_continuous(pv,r,T):
    return pv * math.exp(r * T)
fv = future_value_continuous(100, 0.05, 1)
assert math.isclose(fv, 105.12710963760242)
```

```python
def rate_continuous(fv,pv,T):
    return math.log(fv / pv) / T
r = rate_continuous(105.12710963760242, 100, 1)
assert math.isclose(r, 0.05)
```

```python
t = list(range(31))
pv = 100
r = 0.05
comparison_array = [(time, future_value_discrete(pv, r, time), future_value_continuous(pv, r, time)) for time in t]
print(comparison_array)
```

```python
import matplotlib.pyplot as plt

times = [row[0] for row in comparison_array]
discrete_values = [row[1] for row in comparison_array]
continuous_values = [row[2] for row in comparison_array]

plt.plot(times, discrete_values, label="Discrete")
plt.plot(times, continuous_values, label="Continuous")

plt.xlabel("Time")
plt.ylabel("Future value")
plt.title("Discrete vs. Continuous Future Value")
plt.legend()
plt.grid(True)
plt.show()
```
